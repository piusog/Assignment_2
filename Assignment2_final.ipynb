```python
# Q1.1 — Use zip() and map() to pair names with bills

# Data: (customer_name, units_used, connection_type)
customers = [
    ("Bosco",     245, "Domestic"),
    ("Pius",      520, "Domestic"),
    ("Denis",     410, "Commercial"),
    ("Martha",    802, "Commercial"),
    ("Paul",      150, "Domestic"),
    ("Rachael",   610, "Commercial"),
    ("Peter",     498, "Domestic"),
    ("Karen",     730, "Commercial"),
    ("Charlotte", 305, "Domestic"),
    ("Daniella",  540, "Domestic"),
]

# Rates (UGX per unit)
DOM_RATE = 820
COM_RATE = 1000

# Compute each customer's bill using map + lambda
bills = list(map(
    lambda rec: rec[1] * (DOM_RATE if str(rec[2]).lower() == "domestic" else COM_RATE),
    customers
))

# Pair names with their computed bills using zip
name_bill_pairs = list(zip((r[0] for r in customers), bills))

# Print nicely
print("Q1.1 — Name → Total Bill (UGX):")
for name, bill in name_bill_pairs:
    print(f"{name:<10} : {bill:,} UGX")
```

```python
# 1.2 high consumers
high_consumers = list(filter(lambda rec: rec[1] > 500, customers))
print("\n1.2 High-consumption customers (units > 500):")
for rec in high_consumers:
    print(f"  {rec[0]} — {rec[1]} units ({rec[2]})")
```

```python
# 1.3 *args average
def average_bill(*bills: float) -> float:
    return (sum(bills) / len(bills)) if bills else 0.0

avg_all = average_bill(*bills)
print(f"\n1.3 Average bill across all customers: {avg_all:,.2f} UGX")
```

```python
# 1.4 list comprehension format
formatted = [f"{name}: UGX {bill:,}" for name, bill in name_bill_pairs]
print("\n1.4 Formatted results:")
print(formatted)
```

```python
## Q2: Market Basket Price Aggregator

from itertools import zip_longest  # <-- add this import

mango_prices   = [2500, 2700, 2600, 2800]
orange_prices  = [3000, 3200, 3100, 3050]
apple_prices   = [4500, 4600, 4550, 4700]

# 2.1 zip_longest
zipped_prices = list(zip_longest(mango_prices, orange_prices, apple_prices, fillvalue=None))
print("2.1 zip_longest (rows = vendors):")
for row in zipped_prices:
    print(" ", row)
```

```python
# 2.2 averages via lambda
avg = lambda lst: sum(lst)/len(lst)
avg_mango  = avg(mango_prices)
avg_orange = avg(orange_prices)
avg_apple  = avg(apple_prices)
print(f"\n2.2 Average Mango:  {avg_mango:.2f}")
print(f"2.2 Average Orange: {avg_orange:.2f}")
print(f"2.2 Average Apple:  {avg_apple:.2f}")
```

```python
# 2.3 generator for > 3000
fruit_avgs = {"Mango": avg_mango, "Orange": avg_orange, "Apple": avg_apple}
gen_above_3000 = (fruit for fruit, a in fruit_avgs.items() if a > 3000)
```

```python
# 2.4 print neatly
above_list = list(gen_above_3000)
print("\n2.4 Fruits Above Average Price:", ", ".join(above_list))
```

```python
# Q4 — School Fees Payment Analyzer

students = ["Alex", "Grace", "Sarah", "Brian" ]
installments = [
    [150000, 200000, 250000],
    [500000,      0, 200000],
    [300000, 300000, 300000],
    [400000, 100000,      0]
]

# 4.1 Write a lambda function that sums valid payments and ignores zeros.
sum_valid = lambda pays: sum(filter(lambda x: x > 0, pays))

# 4.2 Use map() to compute total paid by each student.
totals = list(map(sum_valid, installments))   # [600000, 700000, 900000, 500000]

# 4.3 Combine student names and totals using zip().
student_totals = list(zip(students, totals))  # [('Alex',600000), ('Grace',700000), ...]

# 4.4 Use filter() to find students who have cleared full fees (≥ 600,000).
cleared = list(filter(lambda kv: kv[1] >= 600_000, student_totals))

# 4.5 Create payment_summary(**kwargs) that prints summaries like "Alex: 600,000 UGX — Cleared".
def payment_summary(**kwargs):
    """kwargs: e.g., payment_summary(Alex=600000, Grace=700000, ...)
    Prints "<name>: <amount> UGX — Cleared/Pending"
    """
    for name, amount in kwargs.items():
        status = "Cleared" if amount >= 600_000 else "Pending"
        print(f"{name}: {amount:,} UGX — {status}")

# ---- Pretty prints for your submission ----
print("Q4.2 Totals:", totals)
print("\nQ4.3 Student totals (name, total):")
for name, total in student_totals:
    print(f"  {name:<6} : {total:,} UGX")

print("\nQ4.4 Cleared (≥ 600,000):")
for name, total in cleared:
    print(f"  {name:<6} : {total:,} UGX")

print("\nQ4.5 Payment Summary:")
payment_summary(**dict(student_totals))  # pass name=amount pairs
```

```python
## Q5: Agricultural Yield Estimator
# 5.1 Yields in tons
from typing import Dict, List, Tuple

districts   = ['Bushenyi', 'Mityana', 'Kasese', 'Mbale']
yield_data  = [1200, 1500, 900, 1300]   # kg

kg_to_ton = lambda kg: kg / 1000.0
yields_tons = [kg_to_ton(kg) for kg in yield_data]
print("5.1 Yields in tons:", yields_tons)
```

```python
## 5.2 Districts with yield > 1 ton
gen_high = (districts[i] for i, y in enumerate(yields_tons) if y > 1.0)
high_districts = list(gen_high)
print("5.2 Districts with yield > 1 ton:", high_districts)
```

```python
## 5.3 Average yield (tons)
def average_yield_tons(*tons):
    return sum(tons) / len(tons) if tons else 0.0
avg_yield = average_yield_tons(*yields_tons)
print(f"5.3 Average yield (tons): {avg_yield:.3f}")
```

```python
 ## 5.4 Revenue summary
def compute_revenue_tons(prices_per_ton: Dict[str, int]) -> List[Tuple[str, float, int]]:
    results = []
    for d, t in zip(districts, yields_tons):
        price = prices_per_ton.get(d, 0)
        revenue = int(round(t * price))
        results.append((d, t, revenue))
    return results
```

```python
## 5.5 Revenue summary 
prices = {'Bushenyi': 5_000_000, 'Mityana': 4_800_000, 'Kasese': 5_000_000, 'Mbale': 4_900_000}
revenues = compute_revenue_tons(prices)
print("5.4–5.5 Revenue summary:")
for d, t, r in revenues:
    print(f"  {d} produced {t:.1f} tons — Revenue: UGX {r:,}")
```

```python
## Q6: Web Data Aggregation
 
# 6.1 Status codes

sites = ['https://ucu.ac.ug', 'https://harba.ug', 'https://www.bou.or.ug']

def http_status_codes(*urls) -> Dict[str, int]:
    codes = {}
    for u in urls:
        try:
            import requests
            r = requests.get(u, timeout=5)
            codes[u] = r.status_code
        except Exception:
            codes[u] = None
    return codes

codes = http_status_codes(*sites)
print("6.1 Status codes:", codes)
```

```python
# 6.2 Use a list comprehension to print only URLs with status code 200
active_200 = [u for u, c in codes.items() if c == 200]
print("6.2 Reachable (HTTP 200):", active_200)
```

```python
# 6.3 Store results in a dictionary using a dictionary comprehension.
results_dict = {u: (c if c is not None else "No Connection/Timeout") for u, c in codes.items()}
print("6.3 Results dictionary:", results_dict)
```

```python
# 6.4 Create a generator expression that yields 'Active Site: <url>' for reachable domains.
active_gen = (f"Active Site: {u}" for u, c in codes.items() if c == 200)
print("6.4 Active sites:", list(active_gen) or ["None (offline or unreachable here)"])
```
